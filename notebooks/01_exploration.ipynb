```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Load Data
df = pd.read_csv('../data/german_credit_data.csv')

# 1. The Imbalance Check
print("Shape:", df.shape)
counts = df['Class'].value_counts()
print("\nCounts:\n", counts)
print(f"\nFraud Percentage: {counts[1] / len(df):.4%}")

# 2. Visualizing the "Needle"
plt.figure(figsize=(6,4))
sns.countplot(x='Class', data=df)
plt.title("The Imbalance Problem: 0 (Legit) vs 1 (Fraud)")
plt.yscale('log') # Log scale because otherwise you can't even SEE the frauds!
plt.show()
```
